# Training the Regression ModelThis notebook is used for training the regression model using the combined CSV data. It utilizes functions defined in the `ai_model` package for data loading, preprocessing, training, and evaluation.

```python
import pandas as pd
from ai_model.data import load_and_combine_data
from ai_model.preprocess import preprocess_data
from ai_model.features import select_features
from ai_model.train import train_model
from ai_model.evaluate import evaluate_model

# Load and combine the data
data = load_and_combine_data()

# Preprocess the data
cleaned_data = preprocess_data(data)

# Select features for the model
X, y = select_features(cleaned_data)

# Train the regression model
model, X_test, y_test = train_model(X, y)

# Evaluate the model
metrics = evaluate_model(model, X_test, y_test)
metrics
```
